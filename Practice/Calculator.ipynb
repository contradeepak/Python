```python
# Calculator project in Python using ipywidgets.

from ipywidgets import IntText, Button, HTML, VBox 
box1 = IntText(description='Enter a number: ')
box2 = IntText(description='Enter a number: ')
btn = Button(description='Calculate')
out = HTML(value="<b>Result yaha aayega</b>")





def calculate(b):
    num1 = box1.value
    num2 = box2.value
    c = num1 + num2
    d = num1 - num2
    e = num1 * num2
    f = num1 / num2 if num2 != 0 else "Error: Division by zero"
    out.value = f"Addition: {c} <br> Subtraction: {d} <br> Multiplication: {e} <br> Division: {f}"
btn.on_click(lambda x: calculate(None))
VBox([box1, box2, btn, out])
```
